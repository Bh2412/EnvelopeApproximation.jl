# Bessel Functions

```julia
@variables x
```

```julia
sin_taylor = taylor(sin(x), x,0, 0:12)
```

```julia
cos_taylor = taylor(cos(x), x,0, 0:12)
```

```julia
j₀_taylor = simplify(sin_taylor // x)
```

```julia
j₁_taylor = simplify((sin_taylor - x*cos_taylor) // x^2)
```

```julia
j2_taylor = simplify((3 - x ^ 2) * sin_taylor // x ^ 3 - 3 * cos_taylor // x ^ 2)
```

# P_false_vacuum

```julia
@variables u, r
```

```julia
td = u * r
```

```julia
I_ur = exp(u*r//2) + exp(-u*r//2) + ((u^2 - 1)*r//4 - 1) * exp(-r//2)
```

```julia
expansion = Symbolics.taylor(I_ur, r, 0, 0:3)
```

```julia
# A. The Constant Term (c0)
# The most reliable way is to simply evaluate the expression at r=0
c0 = substitute(expansion, Dict(r => 0))

# B. The Linear Term (c1)
c1 = Symbolics.coeff(expansion, r)

# C. The Quadratic Term (c2)
c2 = Symbolics.coeff(expansion, r^2)

# 3. Output / Simplify
println("c0: ", simplify(c0))
println("c1: ", simplify(c1))
println("c2: ", simplify(c2))
```
